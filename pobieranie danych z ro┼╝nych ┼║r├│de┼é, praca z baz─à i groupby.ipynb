````python
def get_station_daily_data(station_key, dni):
    url_path = f"https://powietrze.gios.gov.pl/pjp/current/station_details/table/{station_key}/{dni}/0"
    df = pd.read_html(url_path)[0]
    df = df.iloc[:, :-1]
    df = df.iloc[:-3, :]


    new_col_names = ["_".join(col) for col in df.columns]
    df.columns = new_col_names

    df.insert(loc=0, column='station_key', value=station_key)

    df.rename(columns={df.columns[1]: 'data_godzina'}, inplace=True)

    return df
````

## Tekst na stronie html - wyrażenia regularne - https://pythex.org/
https://powietrze.gios.gov.pl/pjp/content/contact

````python
text = '''
Główny Inspektor Ochrony Środowiska
Joanna Piekutowska
Sekretariat:
tel. 22 36 92 209
e-mail: sekretariatgios@gios.gov.pl



I Zastępca Głównego Inspektora Ochrony Środowiska
Anna Dębowiec
Sekretariat:
tel. 22 36 92 567
e-mail: sekretariatzgios@gios.gov.pl


II Zastępca Głównego Inspektora Ochrony Środowiska
Alicja Żelichowska
Sekretariat:
tel. 22 36 92 567
e-mail: sekretariatzgios@gios.gov.pl


Dyrektor Generalny Głównego Inspektoratu Ochrony Środowiska
Damian Jakubik
Sekretariat:
tel. 22 36 92 630
e-mail: sekretariatdg@gios.gov.pl
'''
````

````python
import re
````

````python
re_phone = r'(\d{2} \d{2} \d{2} \d{3})'
````

````python
re_stanowisko = r'(.+Ochrony Środowiska)'
````

````python
re_imie_nazwisko = r'Ochrony Środowiska\n(.+ .+)\n'
````

````python
re_email = r'e-mail: (.+@.+)'
````

````python
re.findall(re_email, text)
````

````python
re_imie_nazwisko_wer2 = r'\n(\S+ \S+)\n\w'
````

````python
data_dict = {"imie i nazwisko": re.findall(re_imie_nazwisko,text),
             "email": re.findall(re_email, text),
             "stanowisko": re.findall(re_stanowisko,text),
             "phone": re.findall(re_phone,text)
             }
````

````python
df_contacts = pd.DataFrame(data_dict)  # zadanie: utworzyc dataframe z kolumnami: stanowisko, imie_nazwisko, email, telefon
````

````python
df_contacts
````

Skorzystanie z entity extraction: https://spacy.io/

````python
# pip install -U spacy
# python -m spacy download en_core_web_sm
import spacy

# Load English tokenizer, tagger, parser and NER
nlp = spacy.load("en_core_web_sm")

# Process whole documents
text = ("When Sebastian Thrun started working on self-driving cars at "
        "Google in 2007, few people outside of the company took him "
        "seriously. “I can tell you very senior CEOs of major American "
        "car companies would shake my hand and turn away because I wasn’t "
        "worth talking to,” said Thrun, in an interview with Recode earlier "
        "this week.")
doc = nlp(text)

# Analyze syntax
print("Noun phrases:", [chunk.text for chunk in doc.noun_chunks])
print("Verbs:", [token.lemma_ for token in doc if token.pos_ == "VERB"])

# Find named entities, phrases and concepts
for entity in doc.ents:
    print(entity.text, entity.label_)
````

## ładowanie danych godzinowych dla wybranej stacji

````python
def load_station_daily_data(db, station_key, dni, if_exist="append"):
  df = get_station_daily_data(station_key, dni)
  df.to_sql("pomiary_godzinowe", db, if_exists=if_exist, index=False)
````

````python
load_station_daily_data(db, "17160", "3", "replace")  # za pierwszym razem tworzę tabelę
````

````python
load_station_daily_data(db, "9913", "3", "append")  #  za każdym kolejnum razem dodaję wiersze do tabeli
````
